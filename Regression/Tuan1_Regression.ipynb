````python
# Thư viện cần thiết
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
# Tạo dữ liệu ngẫu nhiên
# x là biến đầu vào, y là biến đầu ra
np.random.seed(0)
x = 2 * np.random.rand(100, 1)
y = 4 + 3 * x + np.random.randn(100, 1) # Đường thẳng với nhiễu
# Chia dữ liệu thành tập huấn luyện và tập kiểm tra
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2,
random_state=42)
# Tạo mô hình hồi quy tuyến tính
model = LinearRegression()
# Huấn luyện mô hình trên tập huấn luyện
model.fit(x_train, y_train)
# Dự đoán trên tập kiểm tra
y_pred = model.predict(x_test)
# Đánh giá mô hình
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
# In kết quả
print(f"Hệ số chặn (Intercept): {model.intercept_}")
print(f"Hệ số góc (Slope): {model.coef_}")
print(f"Mean Squared Error (MSE): {mse}")
print(f"R^2 score: {r2}")

# Vẽ biểu đồ
plt.scatter(x_test, y_test, color="blue", label="Dữ liệu thực")
plt.plot(x_test, y_pred, color="red", label="Dự đoán hồi quy")
plt.title("Hồi quy tuyến tính đơn giản")
plt.xlabel("Biến độc lập (x)")
plt.ylabel("Biến phụ thuộc (y)")
plt.legend()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Thiết lập seed để kết quả không đổi
np.random.seed(42)

# 1. Tạo ngẫu nhiên 200 chỉ số BMI từ 15 đến 35
bmi = 15 + 20 * np.random.rand(200, 1)

# 2. Tạo logic mắc bệnh: BMI càng cao, xác suất mắc bệnh càng cao
# Sử dụng hàm sigmoid giả định để chuyển đổi BMI thành xác suất từ 0 đến 1
# Điểm mốc trung bình là BMI = 25
proba = 1 / (1 + np.exp(-(0.5 * (bmi - 25))))

# Quyết định nhãn 1 (Mắc bệnh) hoặc 0 (Khỏe mạnh) dựa trên xác suất trên
mac_benh = (np.random.rand(200, 1) < proba).astype(int).ravel()

# --- KIỂM THỬ DỮ LIỆU VỪA TẠO ---
print("5 mẫu BMI đầu tiên:\n", bmi[:5].ravel())
print("5 nhãn mắc bệnh tương ứng (0: Khỏe, 1: Bệnh):\n", mac_benh[:5])
print(f"Tổng số ca mắc bệnh (nhãn 1): {sum(mac_benh)} / 200")
````

````python
# 1. Chia dữ liệu thành tập huấn luyện (80%) và tập kiểm tra (20%)
X_train, X_test, y_train, y_test = train_test_split(bmi, mac_benh, test_size=0.2, random_state=42)

# 2. Khởi tạo mô hình Hồi quy Logistic
model_logistic = LogisticRegression()

# 3. Huấn luyện mô hình trên tập train
model_logistic.fit(X_train, y_train)

# --- KIỂM THỬ MÔ HÌNH SAU HUẤN LUYỆN ---
print("Hệ số góc (Coefficient):", model_logistic.coef_[0][0])
print("Hệ số chặn (Intercept):", model_logistic.intercept_[0])
print("Giải thích: Hệ số góc dương chứng minh BMI tỷ lệ thuận với khả năng mắc bệnh.")
````

````python
# 1. Dự đoán nhãn lớp (0 hoặc 1)
y_pred = model_logistic.predict(X_test)

# 2. Dự đoán xác suất thực tế (Trả về 2 cột: [Xác suất nhãn 0, Xác suất nhãn 1])
y_prob = model_logistic.predict_proba(X_test)[:, 1]

# 3. Đánh giá hiệu năng
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)

# --- KIỂM THỬ KẾT QUẢ DỰ ĐOÁN TRÊN 5 BỆNH NHÂN ĐẦU TIÊN ---
print("Chỉ số BMI thực tế của 5 người: ", X_test[:5].ravel())
print("Xác suất mắc bệnh dự báo (%):     ", np.round(y_prob[:5] * 100, 2))
print("Nhãn dự đoán của mô hình:          ", y_pred[:5])
print("Nhãn thực tế của bệnh nhân:        ", y_test[:5])
print("\n--- KẾT QUẢ ĐÁNH GIÁ CHUNG ---")
print(f"Độ chính xác (Accuracy): {accuracy * 100:.2f}%")
print("Ma trận nhầm lẫn (Confusion Matrix):\n", conf_matrix)
````

````python
# Tạo một chuỗi liên tục các chỉ số BMI từ 15 đến 35 để vẽ đường cong mượt mà
X_curve = np.linspace(15, 35, 300).reshape(-1, 1)
y_curve_prob = model_logistic.predict_proba(X_curve)[:, 1]

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))

# Vẽ các điểm dữ liệu kiểm tra thực tế
plt.scatter(X_test, y_test, color='blue', alpha=0.7, label='Dữ liệu kiểm tra thực tế (0 hoặc 1)')

# Vẽ đường cong hồi quy Logistic đại diện cho xác suất
plt.plot(X_curve, y_curve_prob, color='red', linewidth=2.5, label='Đường cong xác suất Logistic')

# Vẽ đường ranh giới quyết định (Decision Boundary) tại ngưỡng xác suất 0.5
decision_boundary = -model_logistic.intercept_[0] / model_logistic.coef_[0][0]
plt.axvline(x=decision_boundary, color='green', linestyle='--', label=f'Ranh giới quyết định (BMI ~ {decision_boundary:.2f})')
plt.axhline(y=0.5, color='gray', linestyle=':')

plt.xlabel('Chỉ số khối cơ thể - BMI')
plt.ylabel('Xác suất mắc bệnh')
plt.title('Hồi quy Logistic: Dự đoán khả năng mắc bệnh dựa trên BMI')
plt.legend(loc='lower right')
plt.grid(True, alpha=0.3)
plt.show()

print(f"Kiểm thử ranh giới quyết định: Người có BMI lớn hơn {decision_boundary:.2f} sẽ được dự đoán là Mắc bệnh.")
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Chọn đặc trưng BMI để dự đoán khả năng mắc bệnh (Outcome)
X_col = 'BMI'
y_col = 'Outcome'

if X_col in df_custom.columns and y_col in df_custom.columns:
    # Lấy dữ liệu thực tế
    X_c = df_custom[[X_col]].values
    y_c = df_custom[y_col].values

    # 1. Chia tập dữ liệu thành Train (80%) và Test (20%)
    X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
        X_c, y_c, test_size=0.2, random_state=42
    )

    # 2. Khởi tạo và huấn luyện mô hình hồi quy Logistic
    model_custom = LogisticRegression()
    model_custom.fit(X_train_c, y_train_c)

    # 3. Dự đoán trên tập kiểm tra
    y_pred_c = model_custom.predict(X_test_c)
    y_prob_c = model_custom.predict_proba(X_test_c)[:, 1]

    # 4. Đánh giá hiệu năng
    accuracy_c = accuracy_score(y_test_c, y_pred_c)
    conf_matrix_c = confusion_matrix(y_test_c, y_pred_c)

    print("--- KẾT QUẢ ĐÁNH GIÁ MÔ HÌNH HỒI QUY LOGISTIC TỪ FILE CSV ---")
    print(f"Hệ số góc (Slope): {model_custom.coef_[0][0]:.4f}")
    print(f"Hệ số chặn (Intercept): {model_custom.intercept_[0]:.4f}")
    print(f"Độ chính xác trên tập kiểm tra (Accuracy): {accuracy_c * 100:.2f}%")
    print("Ma trận nhầm lẫn (Confusion Matrix):")
    print(conf_matrix_c)

    # 5. Vẽ đồ thị trực quan hóa
    X_curve = np.linspace(X_c.min(), X_c.max(), 300).reshape(-1, 1)
    y_curve_prob = model_custom.predict_proba(X_curve)[:, 1]

    plt.figure(figsize=(10, 6))
    plt.scatter(X_test_c, y_test_c, color='forestgreen', alpha=0.5, label='Dữ liệu thực tế (Test)')
    plt.plot(X_curve, y_curve_prob, color='red', linewidth=2.5, label='Đường cong xác suất Logistic')

    # Tính ranh giới quyết định (Decision Boundary) tại ngưỡng xác suất 0.5
    db_point = -model_custom.intercept_[0] / model_custom.coef_[0][0]
    plt.axvline(x=db_point, color='darkblue', linestyle='--', label=f'Ranh giới quyết định (BMI ~ {db_point:.2f})')
    plt.axhline(y=0.5, color='gray', linestyle=':')

    plt.xlabel('Chỉ số khối cơ thể (BMI thực tế từ CSV)')
    plt.ylabel('Xác suất mắc bệnh (Outcome)')
    plt.title('Hồi quy Logistic trên file CSV thực tế: BMI vs Outcome')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    print(f"Không tìm thấy các cột cần thiết '{X_col}' hoặc '{y_col}' trong file CSV.")
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Khởi tạo dữ liệu gốc
np.random.seed(42)
x = 30 + 120 * np.random.rand(100, 1)  # Diện tích từ 30m2 đến 150m2
y_true = 100 + 15 * x                 # Phương trình gốc không nhiễu

# Định nghĩa 3 mức độ nhiễu khác nhau (độ lệch chuẩn của nhiễu)
noise_levels = {
    'Nhiễu thấp (Std = 50)': 50,
    'Nhiễu trung bình (Std = 150)': 150,
    'Nhiễu cao (Std = 450)': 450
}

# 2. Vẽ đồ thị so sánh
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharex=True, sharey=True)

for i, (label, std) in enumerate(noise_levels.items()):
    # Tạo nhiễu ngẫu nhiên
    noise = np.random.randn(100, 1) * std
    y_noisy = y_true + noise

    # Huấn luyện mô hình hồi quy tuyến tính
    model = LinearRegression()
    model.fit(x, y_noisy)
    y_pred = model.predict(x)

    # Tính toán chỉ số đánh giá
    mse = mean_squared_error(y_noisy, y_pred)
    r2 = r2_score(y_noisy, y_pred)

    # Vẽ biểu đồ tương ứng
    axes[i].scatter(x, y_noisy, color='blue', alpha=0.6, label='Dữ liệu thực tế')
    axes[i].plot(x, y_pred, color='red', linewidth=2, label='Đường dự đoán')
    axes[i].set_title(f"{label}\nMSE: {mse:.1f} | R²: {r2:.4f}")
    axes[i].set_xlabel('Diện tích (m2)')
    if i == 0:
        axes[i].set_ylabel('Giá bán (Triệu VND)')
    axes[i].legend()
    axes[i].grid(True, alpha=0.3)

plt.suptitle('So sánh tác động của các mức độ nhiễu lên mô hình Hồi quy tuyến tính', fontsize=16, y=1.05)
plt.tight_layout()
plt.show()
````

### **BÀI 3 VỀ NHÀ: HỒI QUY TUYẾN TÍNH VỚI TẬP DỮ LIỆU GIÁ NHÀ**

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Tải bộ dữ liệu California Housing
housing = fetch_california_housing(as_frame=True)
df_housing = housing.frame

print("Danh sách các thuộc tính đặc trưng (Features):")
print(housing.feature_names)
print("\n5 dòng dữ liệu đầu tiên:")
display(df_housing.head())

# Chọn đặc trưng 'MedInc' (Thu nhập trung bình khu vực) làm biến dự đoán giá nhà 'MedHouseVal'
X_house = df_housing[['MedInc']].values
y_house = df_housing['MedHouseVal'].values
````

````python
# 2. Phân chia dữ liệu thành tập huấn luyện (80%) và tập kiểm tra (20%)
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(
    X_house, y_house, test_size=0.2, random_state=42
)

# 3. Huấn luyện mô hình hồi quy tuyến tính
model_house = LinearRegression()
model_house.fit(X_train_h, y_train_h)

# 4. Dự đoán trên tập kiểm tra
y_pred_h = model_house.predict(X_test_h)

# 5. Tính toán các chỉ số đánh giá
mse_h = mean_squared_error(y_test_h, y_pred_h)
r2_h = r2_score(y_test_h, y_pred_h)

print("--- KẾT QUẢ ĐÁNH GIÁ MÔ HÌNH DỰ ĐOÁN GIÁ NHÀ ---")
print(f"Hệ số góc (Slope): {model_house.coef_[0]:.4f}")
print(f"Hệ số chặn (Intercept): {model_house.intercept_:.4f}")
print(f"Mean Squared Error (MSE): {mse_h:.4f}")
print(f"R2 Score (Độ tương quan giải thích): {r2_h * 100:.2f}%")
````

````python
# 6. Trực quan hóa kết quả dự đoán thực tế
plt.figure(figsize=(10, 6))
plt.scatter(X_test_h[:200], y_test_h[:200], color='teal', alpha=0.6, label='Thực tế (200 mẫu đầu)')
plt.plot(X_test_h[:200], y_pred_h[:200], color='crimson', linewidth=3, label='Đường hồi quy dự báo')
plt.xlabel('Thu nhập trung bình khu vực (MedInc)')
plt.ylabel('Giá trị nhà trung bình (100k USD)')
plt.title('Dự đoán giá nhà California dựa trên Thu nhập trung bình')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````
